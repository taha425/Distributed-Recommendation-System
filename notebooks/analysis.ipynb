# Distributed Recommendation System with Clickstream Analysis
## Big Data Analysis Mini Project Notebook

This notebook provides exploratory data analysis (EDA) and model evaluation for the PySpark MLlib ALS collaborative filtering model and co-purchase graph analytics.

```python
import os
import json
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use('dark_background')
sns.set_palette('crest')
print('Libraries imported successfully!')
```

### 1. Load Clickstream & Model Statistics

```python
processed_dir = os.path.join('..', 'data', 'processed')

with open(os.path.join(processed_dir, 'clickstream_stats.json')) as f:
    clickstream_stats = json.load(f)

with open(os.path.join(processed_dir, 'model_metrics.json')) as f:
    model_metrics = json.load(f)

with open(os.path.join(processed_dir, 'graph_metrics.json')) as f:
    graph_metrics = json.load(f)

print('Pipeline metrics loaded successfully.')
```

### 2. Event Type Distribution

```python
event_df = pd.DataFrame(list(clickstream_stats['event_distribution'].items()), columns=['Event Type', 'Count'])

plt.figure(figsize=(8, 4))
sns.barplot(data=event_df, x='Event Type', y='Count', palette='viridis')
plt.title('Clickstream Event Distribution')
plt.ylabel('Event Count')
plt.show()
```

### 3. Conversion Funnel Analysis

```python
funnel = clickstream_stats['conversion_funnel']
funnel_df = pd.DataFrame([
    {'Stage': 'Total Sessions', 'Count': funnel['total_sessions']},
    {'Stage': 'Add to Cart', 'Count': funnel['sessions_with_add_to_cart']},
    {'Stage': 'Purchase', 'Count': funnel['sessions_with_purchase']}
])

plt.figure(figsize=(8, 4))
sns.barplot(data=funnel_df, x='Stage', y='Count', palette='magma')
plt.title('Session Conversion Funnel')
plt.show()
```

### 4. PageRank vs Degree Centrality (GraphX Analysis)

```python
top_pr_df = pd.DataFrame(graph_metrics['top_products_by_pagerank'])

plt.figure(figsize=(8, 5))
sns.scatterplot(data=top_pr_df, x='degree_centrality', y='pagerank', hue='community_id', palette='Spectral', s=100)
plt.title('Product Authority: PageRank vs Degree Centrality')
plt.xlabel('Degree Centrality')
plt.ylabel('PageRank Score')
plt.show()
```
